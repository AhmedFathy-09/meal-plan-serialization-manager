```python
# Final Project
class Recipe:
    def __init__(self, name, ingredients, prep_time, calories):
        self.name = name
        self.ingredients = ingredients
        self.prep_time = prep_time
        self.calories = calories
class MealPlan:
    def __init__ (self, planName, planID):
        self.planName = planName
        self.planID = planID
        self.recipes = []
    def add_recipe(self, recipe):
        self.recipes.append(recipe)
    def save_mealplan_to_file(self, filename):
        with open(filename,'a')as file:
            for r in self.recipes:
                str_ingredients = ",".join(r.ingredients)
                file.write(f"{r.name}/{str_ingredients}/{r.prep_time}/{r.calories}\n")
    def load_mealplan_from_file(self, filename):
        try:
            with open(filename, 'r')as file:
                for line in file:
                    info = line.strip().split('/')
                    if len(info) != 4:
                        continue
                    name = info[0]
                    ingredients = info[1].split(',')
                    prep_time = int(info[2])
                    calories = int(info [3])
                    new_recipe = Recipe(name, ingredients, prep_time, calories)
                    self.add_recipe(new_recipe)
        except FileNotFoundError:
            print(f"File {filename} not found")
class RecipeBook:
    def __init__ (self):
        self.mps = []
    def add_mealplan(self, mealplan):
        self.mps.append(mealplan)
    def display_mealplans(self):
        if len(self.mps) == 0:
            print("No mealplans found")     
        for m in self.mps:
            print(f"Plan: {m.planName} ID: {m.planID}")
            for r in m.recipes:
                print(f"{r.name}: {r.calories}kcal")
    def search_mealplan(self, name):
        lower_name = name.lower()
        for c in self.mps:
            if c.planName.lower() == lower_name:
                return c
        return None
def testrun():
    recipes = []
    for i in range(1, 4):
        print("Enter Recipe", i)
        name = input("Names: ")

        ingredients_input = input("Ingredients (comma separated): ")
        ingredients = ingredients_input.split(",")

        prep_time = int(input("Prep time (minutes): "))
        calories = int(input("Calories: "))

        recipe = Recipe(name, ingredients, prep_time, calories)
        recipes.append(recipe)
        print("======================================")

    mealplan1 = MealPlan("Plan One", 1)
    mealplan2 = MealPlan("Plan Two", 2)

    mealplan1.add_recipe(recipes[0])
    mealplan1.add_recipe(recipes[1])
    mealplan2.add_recipe(recipes[2])

    filename = 'meal_plan.txt'  
    with open(filename, 'w')as file:
        pass
    mealplan1.save_mealplan_to_file(filename)
    print("Saved Meal Plan One to file:", filename)
    print("======================================")

    book = RecipeBook()

    book.add_mealplan(mealplan1)
    book.add_mealplan(mealplan2)

    print("Displaying all meal plans:\n")
    book.display_mealplans()

    search_name = input("\nEnter meal plan name to search (Plan one or Plan two): ")
    result = book.search_mealplan(search_name)

    if result is not None:
        print("\nMeal plan found:\n")
        print(result.planName)
    else:
        print("\nMeal plan not found.")
testrun()
```
